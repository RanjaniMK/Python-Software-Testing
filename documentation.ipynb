```python
def add_numbers(a,b):
    
    """
    _summary_ : Add two numbers together and return the result.

    Returns:
        _type_: This function takes two numeric values and returns their sum.
                It works with integers, floats, and other numeric types that support
                the addition operator.
    """
    return a+b


print(add_numbers(10.5,80.0))
```
