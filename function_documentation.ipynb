```python
def multiply(x : int, y : int) -> int:
    """_summary_ : Multiplies two numbers and returns the result.

    Args:
        x (int): It is the first number (operand), which is the input.
        y (int): It is the second number(operand), which is the input.

    Returns:
        int: The result returned after executing the above function is of type int.
    """
    return x*y

print(multiply(50,50))
```
